QUES 1. TRY TO BUILD A CLASSIFIER FOR THE MNIST DATASET THAT ACHIEVES OVER 97% ACCURACY ON TEST SET

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
```

```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml("mnist_784", as_frame=False)
```

```python
X,y = mnist.data, mnist.target
```

```python
X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]
```

```python
knn_clf = KNeighborsClassifier()
knn_clf.fit(X_train,y_train)
```

```python
baseline_accuracy = knn_clf.score(X_test,y_test)
```

```python
baseline_accuracy
```

```python
param_grid = [{"weights": ["uniform", "distance"], 'n_neighbors': [3,4,5,6]}]

knn_clf = KNeighborsClassifier()
grid_search = GridSearchCV(knn_clf, param_grid, cv=5)
grid_search.fit(X_train[:10_000], y_train[:10_000])
```

```python
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
grid_search.best_estimator_.fit(X_train,y_train)
tuned_accuracy = grid_search.score(X_test,y_test)
tuned_accuracy
```

QUES 2. WRITE A FUNCTION THAT CAN SHIFT AN MNIST IMAGE IN ANY DIRECTION. THEN FOR EACH IMAGE IN TRAIN SET, CREATE FOUR SHIFTED COPIES AND ADD THEM TO THE TRAINING SET.

```python
from scipy.ndimage import shift
```

```python
def shift_image(image, dx, dy):
    image = image.reshape((28,28))
    shifted_image = shift(image, [dy, dx], cval=0, mode="constant")
    return shifted_image.reshape([-1])
```

```python
import matplotlib.pyplot as plt
image = X_train[1000]
shifted_image_down = shift_image(image, 0, 5)
shifted_image_left = shift_image(image, -5, 0)

plt.figure(figsize=(12,3))
plt.subplot(131)
plt.title("Original")
plt.imshow(image.reshape(28,28), interpolation="nearest", cmap="Greys")
plt.subplot(132)
plt.title("Shifted down")
plt.imshow(shifted_image_down.reshape(28,28), interpolation="nearest", cmap="Greys")
plt.subplot(133)
plt.title("Shifted left")
plt.imshow(shifted_image_left.reshape(28,28), interpolation="nearest", cmap="Greys")
```

```python
import numpy as np
X_train_augmented = [image for image in X_train]
y_train_augmented = [label for label in y_train]

for dx, dy in ((-1,0), (1, 0), (0, 1), (0, -1)):
    for image, label in zip(X_train, y_train):
        X_train_augmented.append(shift_image(image, dx, dy))
        y_train_augmented.append(label)

X_train_augmented = np.array(X_train_augmented)
y_train_augmented = np.array(y_train_augmented)
```

```python
shuffle_idx = np.random.permutation(len(X_train_augmented))
X_train_augmented = X_train_augmented[shuffle_idx]
y_train_augmented = y_train_augmented[shuffle_idx]
```

```python
knn_clf = KNeighborsClassifier(**grid_search.best_params_)
```

```python
knn_clf.fit(X_train_augmented,y_train_augmented)
```

```python
augmented_accuracy = knn_clf.score(X_test, y_test)
```

```python
augmented_accuracy
```

```python
error_rate_change = (1 - augmented_accuracy) / (1 - tuned_accuracy) - 1
print(f"error_rate_change = {error_rate_change: .0%}")
```
